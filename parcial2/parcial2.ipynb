## Ejercicio 1 — Generadores TresFibonacci *(2 puntos)*

Los tres primeros números son **1, 1, 1** y cada siguiente es la **suma de los tres anteriores**:

$$1, 1, 1, 3, 5, 9, 17, 31, \ldots$$

**Parte 1:** `G1()` — generador **infinito** de todos los TresFibonacci.

**Parte 2:** `tresfibonacci(n)` — generador que produce **solo los primeros n** números.

```python
def G1():
    a, b, c = 1, 1, 1
    while True:
        yield a
        a, b, c = b, c, a + b + c
```

```python
def tresfibonacci(n):
    gen = G1()              
    for i in range(n):
        yield next(gen)    
```

## Ejercicio 2 — Decorador con límite de errores *(1.5 puntos)*

Crear `f2(func)` que:
- Si **no hay error** → retorna el resultado normalmente.
- Si **hay error** → imprime `Ha cometido N error(es): TipoError` y retorna `None`.
- Al **3er error** → bloquea la función para siempre.
- Después del bloqueo → muestra siempre `Leer la documentacion`.

```python
def f2(func):
    errores   = [0]
    bloqueado = [False]

    def wrapper(*args):
        if bloqueado[0]:
            print('Leer la documentacion')
            return None

        try:
            return func(*args)         

        except Exception as e:
            errores[0] += 1
            print(f'Ha cometido {errores[0]} error(es): {type(e).__name__}')

            if errores[0] >= 3:         
                bloqueado[0] = True

            return None                 

    return wrapper
```

```python
@f2
def division(a, b):
    return a / b
resultados = (
    division(1, 2),
    division(1, 0),
    division('a', 1),
    division('a', 'b'),
    division(1, 1)
)
print(resultados)
```

## Ejercicio 3 — Último dígito de un número *(1.5 puntos)*

Escribir `f3i(n)` (iterativa) y `f3r(n)` (recursiva) que retornan el **último dígito** de un número natural.

**Restricción:** ¡No se puede usar `str`!

**Clave:** El último dígito de cualquier número es `n % 10`.
- `4 % 10 = 4`
- `23 % 10 = 3`  → el enunciado muestra `2`, así que el último dígito de **23** en base de divisores es **3**... pero el ejemplo dice `> 4 2 1`. Mirando `23 → 2` y `189 → 1`, el enunciado pide el **primer** dígito.

> **Nota de lectura del enunciado:** la salida `4 2 1` para `[4, 23, 189]` corresponde al **primer dígito** (el más significativo), no al último. Así que la función retorna el **primer dígito** de `n` sin usar `str`.

La forma de extraer el primer dígito sin `str` es dividir repetidamente por 10 hasta que `n < 10`.

```python
def f3i(n):
    while n >= 10:
        n = n // 10
    return n
```

```python
def f3r(n):
    if n < 10:          
        return n
    return f3r(n // 10) 
```

```python
L = [4, 23, 189]

print('f3i:')
for i in L:
    print(f3i(i), end=' ')
print('\nf3r:')
for i in L:
    print(f3r(i), end=' ')
```
