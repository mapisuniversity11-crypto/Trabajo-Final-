1. Una función de nombre `f1(f, a, b)` recibe una función f y dos números entros a, b. Retorna la sumatoria $$\sum_{a}^{b} f(i)\cdot i $$ Veamos un ejemplo:
```python
f1(lambda x: x**2, 1, 3)
> 36 # 1^2 * 1 + 2^2 * 2 + 3^3 * 3
```

````python
# Una función de nombre f1 recibe una función y dos números enteros a,b
# Retorna la sumatoria f(i)*i para los números enteros i entre a y b

def f1(func, a, b):
    ResultadoFinal = 0
    for i in range(a, b + 1):
        ResultadoFinal += func(i) * i
    return ResultadoFinal

f1(lambda x: x**2, 1, 3)
````

2. Una función `f2(L)` recibe una lista de números y retorna una función `f(x)` que representa el polinomio con coeficientes en `L` evaluado en `x`. Veamos un ejemplo:
```python
# Ejemplo de uso: si L = [1, 2, 3], entonces f(x) = 1 + 2*x + 3*x^2
polinomio = f2([1, 2, 3])
polinomio(2)
> 17
```

````python
# Una función f2(L) recibe una lista de números y retorna una función f(x)
# que representa el polinomio con coeficientes en L evauado en X

def f2(L):
    def f(x):
        Result = 0
        for i in range(len(L)):
            Result += L[i] * (x ** i)
        return Result
    return f

# Ejemplo de uso: si L = [1, 2, 3], entonces f(x) = 1 + 2*x + 3*x^2
polinomio = f2([1, 2, 3])
print(polinomio(2)) 
````

3. Una función `f3(x0, y0)` recibe dos números y retorna dos funciones `recta(x)` y `paralela(x)`. La función `recta(x)` representa la recta que pasa por `(x0, y0)` y tiene pendiente 2.La función `paralela(x)` representa la recta paralela a la recta anterior que pasa por `(1,1)`. Veamos un ejemplo de uso
```python
recta, paralela = f3(2, 2) 
recta(3), paralela(3)  # Evaluar la recta en x=3, evaluar la recta paralela en x=3
> (4,5)
```


````python
# Una función f3(x0, y0) recibe dos números y retorna dos funciones recta(x) y paralela(x)
# La función recta representa la recta que pasa por (x0, y0) y tiene pendiente 2
# La función paralela(x) representa la recta paralela a la recta anterior que pasa por (1,1)
def f3(x0, y0):
    def recta(x):
        return 2 * (x - x0) + y0  # Ecuación de la recta con pendiente 2 que pasa por (x0, y0)
    
    def paralela(x):
        return 2 * (x - 1) + 1 
    return recta, paralela
recta, paralela = f3(2, 2)
recta(3), paralela(3) 
````

4. Una función f4(L) recibe una lista de números enteros y debe retornar la suma de aquellos 
números mayores que 10 y cuya suma del primer y último dígito sea par. Veamos un ejemplo
```python
f4([12, 15, 23, 34, 45, 56, 11])  # Debería retornar 15 + 11 = 26
> 26
```

````python
# Una función f4(L) recibe una lista de números enteros y debe retornar la suma de aquellos 
# números mayores que 10 y cuya suma del primer y último dígito sea par.
def f4(L):
    L = [x for x in L if x > 10 and (int(str(x)[0]) + int(str(x)[-1])) % 2 == 0]
    return sum(L)

print(f4([12, 15, 23, 34, 45, 56, 11]))
````
