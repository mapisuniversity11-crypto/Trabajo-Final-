# 🏦 Punto 1: Clase `Cuenta`

Dicha cuenta debe recrear una cuenta bancaria básica que debe tener los siguientes métodos:

## Métodos requeridos:

* `__init__(self, titular, saldo)`: Dichac cuenta se inicia con un `titular` y un saldo inicial `saldo`.
* `depositar(self, cantidad)`: Permite depositar una `cantidad` y debe modificar el saldo. 
    * *Nota: Asegurarse que la cantidad depositada sea un número positivo.*
* `retirar(self, cantidad)`: Permite retirar de la cuenta, pero debe asegurarse de tener los fondos suficientes y modificar el saldo.
* `consultar_saldo(self)`: Permite ver el saldo de la cuenta.

---

### 🚀 Reto final:
**¿Cómo podemos asegurarnos de que el saldo inicial sea un número positivo al iniciar la clase?**


````python
class Cuenta:
    def __init__(self, titular, saldo):
      self.titular = titular
      if saldo >= 0:
        self.saldo = saldo

    def depositar(self, cantidad):
        if cantidad <= 0:
            print("Ingresar una cantidad positiva")
        else:
            self.saldo += cantidad

    def retirar(self, cantidad):
        if 0 < cantidad <= self.saldo:
            self.saldo -= cantidad
        else:
            print("Fondos insuficientes o cantidad no válida.")

    def consultar_saldo(self):
        return self.saldo

c = Cuenta("Luis", 1000)

c.depositar(500)
c.retirar(300)

print(c.consultar_saldo())
````

# 📐 Punto 2: Clase `VectoresR3`

Dicha clase debe representar los vectores usuales en $\mathbb{R}^3$ y contener los siguientes métodos:

## Métodos Especiales:

* `__init__(self, L)`: Se inicia el vector con una lista de tres componentes (ej. `[x, y, z]`).
* `__add__(self, v1)`: Sobrecarga del operador `+` para sumar dos vectores.
* `__rmul__(self, v1)`: Para recrear la multiplicación por un escalar (número por vector).
* `__sub__(self, v1)`: Sobrecarga del operador `-` para restar vectores.

## Métodos de Álgebra Lineal:

* `producto_punto(self, v1)`: Para recrear el producto punto de dos vectores. **Debe retornar un número.**
* `producto_cruz(self, v1)`: Para recrear el producto cruz de dos vectores. **Debe retornar un nuevo vector.**

````python
class lineal:
    def __init__(self, L):
        self.x = L[0]
        self.y = L[1]
        self.z = L[2]
    def __add__(self, v1):
        # Sumamos componente a componente
        return lineal([self.x + v1.x, self.y + v1.y, self.z + v1.z])

    def __sub__(self, v1):
        # Restamos componente a componente
        return lineal([self.x - v1.x, self.y - v1.y, self.z - v1.z])

    def __rmul__(self, escalar):
        # Multiplicación escalar: el número por cada componente
        return lineal([escalar * self.x, escalar * self.y, escalar * self.z])
````
